# 00 · Extracción de datos

Descarga los datos de la competencia desde Kaggle y los deja en `../data/raw/`.

Requisitos previos (una sola vez por persona):

1. Aceptar las reglas en https://www.kaggle.com/competitions/rogii-wellbore-geology-prediction/rules
2. Credenciales: `uv run kaggle auth login` o `KAGGLE_API_TOKEN` en `../.env` (ver README del repo).

Equivalente en terminal: `make data`.

## 1. Descarga (sólo si `data/raw/train` no existe)

```python
from pathlib import Path
import subprocess

RAW = Path("../data/raw")
if not (RAW / "train").exists():
    subprocess.run(["bash", "../scripts/download_data.sh"], check=True)
else:
    print("Datos ya presentes en", RAW.resolve())
```

## 2. Verificación

```python
n_train = len(list((RAW / "train").glob("*__horizontal_well.csv")))
n_test = len(list((RAW / "test").glob("*__horizontal_well.csv")))
print(f"train: {n_train} pozos | test: {n_test} pozos")
assert n_train == 773 and n_test == 3, "conteo inesperado de pozos"
print("archivos extra:", [p.name for p in RAW.iterdir() if p.is_file() and p.name != ".gitkeep"])
```
