**Objetivo:** usar la **señal de gamma ray** para ajustar la trayectoria: el GR del pozo horizontal, proyectado en TVT, debe reproducir la firma del GR del typewell (y la del propio pozo antes del PS). A partir de la trayectoria base de la parte 2, buscar la secuencia de TVT que mejor alinea el GR, combinar ambas fuentes y producir la predicción final.

Referencia: notebook `01_exploracion.ipynb`, secciones 2, 4 y 9 (correlación mediana 0.84 entre GR horizontal y typewell; GR del tramo conocido como mejor referencia que el typewell).

Convenciones comunes a las tres partes:

- Rutas relativas a la raíz del repo: `../data/raw/`, `../data/processed/`, `../models/`, `../submissions/`.
- Todo lo que una parte produce para otra se guarda en `../data/processed/<parte>/` en formato parquet, con las columnas documentadas en la sección *Salidas* de cada notebook.
- Identificador de pozo: el prefijo de 8 caracteres del archivo (`000d7d20`). Índice de punto: la fila del CSV horizontal (el mismo que usa `sample_submission`).
- Métrica oficial: RMSE sobre los puntos con índice ≥ PS (Prediction Start), en pies.

## Alcance

1. **Referencia de GR en TVT**: construir la curva de referencia GR(TVT) para cada pozo combinando el typewell remuestreado y el GR del propio pozo antes del PS (mayor resolución). Definir cómo se ponderan.
2. **Alineación secuencial**: a lo largo del tramo a predecir, estimar en cada ventana el desplazamiento de TVT que maximiza la similitud entre el GR observado y la referencia. Opciones a evaluar: correlación por ventanas deslizantes con restricciones de suavidad, DTW restringido, filtro de partículas sobre la pendiente de TVT. Usar `tvt_pred` y `tvt_sigma` de la parte 2 como prior.
3. **Manejo de huecos de GR**: en tramos con `GR_mask` largo, la alineación no aporta y debe caer sobre la trayectoria base.
4. **Combinación y suavizado**: fusionar trayectoria geométrica y alineación de GR; imponer suavidad (las curvas de TVT reales son suaves). Opcional: modelo de residuo (1D-CNN / U-Net sobre secuencias, o tabular) entrenado out-of-fold con las dos predicciones como entradas.
5. **Predicción final y diagnóstico**: RMSE out-of-fold global, por fold y por pozo; figuras de los peores pozos; predicción para los pozos de `test/` en formato `sample_submission`.
